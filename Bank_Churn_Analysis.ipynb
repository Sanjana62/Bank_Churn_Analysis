```python
import pandas as pd
df = pd.read_csv("BankChurners.csv")
df.info()
df.describe()
df["Attrition_Flag"].value_counts()
df.groupby("Income_Category")["Attrition_Flag"].value_counts(normalize=True)
df.groupby("Card_Category")["Attrition_Flag"].value_counts(normalize=True)
df.groupby("Marital_Status")["Attrition_Flag"].value_counts(normalize=True)
import seaborn as sns
import matplotlib.pyplot as plt

# 1. Churn rate by income category
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Income_Category", y=(df["Attrition_Flag"]=="Attrited Customer").astype(int), estimator="mean")
plt.ylabel("Churn Rate")
plt.title("Churn Rate by Income Category")
plt.xticks(rotation=45)
plt.show()

# 2. Churn rate by card category
plt.figure(figsize=(8,5))
sns.barplot(data=df, x="Card_Category", y=(df["Attrition_Flag"]=="Attrited Customer").astype(int), estimator="mean")
plt.ylabel("Churn Rate")
plt.title("Churn Rate by Card Category")
plt.show()

# 3. Age distribution split by churn status
plt.figure(figsize=(8,5))
sns.histplot(data=df, x="Customer_Age", hue="Attrition_Flag", kde=True, element="step")
plt.title("Age Distribution by Churn Status")
plt.show()
```
